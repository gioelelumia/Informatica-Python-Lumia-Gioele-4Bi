# Come consegnare un esercizio su GitHub

**Per chi è.** Per un compagno della 4Bi che non ha mai consegnato un esercizio.

**A cosa serve.** A portare un file dal tuo computer a GitHub in quattro passi. Per ogni passo scrivo perché c'è e cosa succede se lo salti. Come esempio uso l'esercizio 3: il file `M1_markdown/risorse.md` e l'immagine `M1_markdown/img/anteprima-risorse.png`.

## Passo 1 – Salva il file e guarda l'anteprima

Git guarda solo i file salvati sul disco. Se accanto al nome del file c'è un pallino, premi `Ctrl+S`. L'anteprima (`Ctrl+K`, poi `V`) serve perché un errore di Markdown non dà messaggi: un recinto aperto, per esempio, trasforma tutto il resto in codice. Se salti questo passo, consegni un file vecchio o reso male e lo scopri solo quando lo corregge il docente.

```bash
git status
```

````python
passi = 4
print("Passi della procedura:", passi)
````

## Passo 2 – Prepara i file con `git add`

Un commit contiene solo i file che hai preparato tu. Così consegni un esercizio alla volta. Se salti questo passo, il commit non salva niente e Git risponde «no changes added to commit».

```bash
git add M1_markdown/risorse.md M1_markdown/img/anteprima-risorse.png
```

````python
file_markdown = 1
file_immagini = 1
file_nel_commit = file_markdown + file_immagini
print("File nel commit:", file_nel_commit)
````

## Passo 3 – Fai il commit

Il commit è la foto del lavoro che resta nella cronologia, e il messaggio è l'unica spiegazione che legge il docente. Per il corso la prima riga ha al massimo cinquanta caratteri, è al presente e non ha il punto finale. Se salti il commit, il push dopo non ha niente da inviare.

```bash
git commit -m "Aggiunge la pagina delle risorse (M1 es. 3)"
```

## Passo 4 – Fai il push

Finché non fai il push, il commit esiste solo sul tuo computer e il docente non lo vede. Se salti questo passo, l'esercizio risulta non consegnato anche se in locale è perfetto. Alla fine controlla su GitHub che il file e l'immagine ci siano.

```bash
git push
```

````python
minuti_salvataggio = 10
minuti_add = 1
minuti_commit = 2
minuti_push = 1
minuti_totali = minuti_salvataggio + minuti_add + minuti_commit + minuti_push
print("Minuti stimati:", minuti_totali)
print("Minuti medi per passo:", minuti_totali / passi)
````

## Riepilogo dei comandi

| Comando | Cosa fa | Se lo salti |
|---|---|---|
| `git status` | mostra i file modificati | non vedi un file dimenticato |
| `git add` | prepara i file per il commit | il commit non salva niente |
| `git commit -m` | salva il lavoro con un messaggio | il push non ha niente da inviare |
| `git push` | manda i commit su GitHub | il docente non vede il lavoro |

## Regola prima di consegnare

> Prima di ogni consegna guarda l'anteprima e leggi `git status`. Se hai fatto un notebook, esegui anche **Kernel > Restart Kernel and Run All Cells**. Se qualcosa non torna, non fare il commit.
