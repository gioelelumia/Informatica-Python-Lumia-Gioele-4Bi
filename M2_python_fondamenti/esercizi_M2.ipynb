# Esercizio 1 — Durata di una sessione di laboratorio

```python
# Programma per convertire i secondi di una sessione in formato HH:MM:SS, ore decimali e controllo limite.

secondi_totali = 15000

ore = secondi_totali//3600
minuti = (secondi_totali%3600)//60
secondi = secondi_totali%60

ore_decimali = secondi_totali / 3600

print(f"{ore:02d}:{minuti:02d}:{secondi:02d}")
print(f"in ore decimali: {ore_decimali:.2f}")
if secondi_totali > 15000:
    print("attenzione: la sessione supera le quattro ore")
```

# Esercizio 2 — Media dei voti e giudizio

```python
# Programma che calcola la media di tre voti e determina il giudizio in base a voti e assenze.

nome_studente = input("inserisci il nome dello studente: ")
primo_voto = float(input("inserisci il primo voto: "))
secondo_voto = float(input("inserisci il secondo voto: "))
terzo_voto = float(input("inserisci il terzo voto: "))
ore_assenza = int(input("inserisci le ore di assenza: "))
media = (primo_voto+secondo_voto+terzo_voto)/3

if ore_assenza > 50:
    giudizio = "non valutabile"
elif media >= 8:
    giudizio = "ottimo"
elif media >= 6:
    giudizio = "sufficente"
else:   
    giudizio = "insufficente"

print (f"{nome_studente}: media {media}, giudizio {giudizio}")
```
