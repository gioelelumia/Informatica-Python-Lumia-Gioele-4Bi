# Esercizio 1 — Durata di una sessione di laboratorio

```python
# Programma per convertire i secondi di una sessione in formato HH:MM:SS, ore decimali e controllo limite.

secondi_totali = 15000

ore = secondi_totali//3600
minuti = (secondi_totali%3600)//60
secondi = secondi_totali%60

ore_decimali = secondi_totali / 3600

print(f"{ore:02d}:{minuti:02d}:{secondi:02d}")
print(f"in ore decimali: {ore_decimali:.2f}")
if secondi_totali > 15000:
    print("attenzione: la sessione supera le quattro ore")
```

# Esercizio 4 — Lettura di un codice di postazione

```python
# Programma che ripulisce un codice postazione ed estrae laboratorio, calcolatore, anno e metadati

codice_input = input("Inserisci il codice di postazione: ")

codice_pulito = codice_input.strip().upper()

laboratorio = codice_pulito[0:4]
calcolatore = codice_pulito[5:9]
anno = codice_pulito[10:14]

lunghezza = len(codice_pulito)
ultimo_carattere = codice_pulito[-1]

print(f"laboratorio: {laboratorio}")
print(f"calcolatore: {calcolatore}")
print(f"anno: {anno}")
print(f"lunghezza: {lunghezza}, ultimo carattere: {ultimo_carattere}")
```

# Esercizio 5 — Conteggio dei caratteri di una frase

```python
# Programma che legge una frase e conta vocali, consonanti, cifre e altri caratteri con un unico ciclo for.
# Come da specifica, le lettere accentate vengono considerate consonanti per semplificazione.

frase = input("inserisci la frase da analizzare: ")

conteggio_vocali = 0
conteggio_consonanti = 0
conteggio_cifre = 0
conteggio_altri = 0

vocali = "aeiou"
consonanti = "bcdfghjklmnpqrstvwxyz"

for carattere in frase:
    carattere_minuscolo = carattere.lower()
    
    if carattere_minuscolo in vocali:
        conteggio_vocali += 1
    elif carattere_minuscolo in consonanti or carattere in "àèéìòù":
        conteggio_consonanti += 1
    elif carattere.isdigit():
        conteggio_cifre += 1
    else:
        conteggio_altri += 1

caratteri_totali = len(frase)
percentuale_vocali = (conteggio_vocali / caratteri_totali) * 100

print(f"vocali: {conteggio_vocali}, consonanti: {conteggio_consonanti}, cifre: {conteggio_cifre}, altri: {conteggio_altri}")
print(f"percentuale di vocali: {percentuale_vocali:.1f}%")
```
